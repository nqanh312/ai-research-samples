```python
from sklearn.metrics import accuracy_score

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    if isinstance(logits, tuple):
        logits = logits[0]

    preds = logits.argmax(axis=-1)
    acc = accuracy_score(labels, preds)
    return {"accuracy": acc}
```

```python
import torch
import torch.nn as nn
import pandas as pd
import os
from transformers import AutoModelForSequenceClassification, AutoTokenizer, Trainer, TrainingArguments
from datasets import Dataset
import time

# Thêm list lưu thời gian huấn luyện để dùng cho đánh giá
train_times = {}

# Định nghĩa các loss function
class FocalLoss(nn.Module):
    def __init__(self, gamma=2.0, alpha=0.25, reduction='mean'):
        super(FocalLoss, self).__init__()
        self.gamma = gamma
        self.alpha = alpha
        self.reduction = reduction

    def forward(self, inputs, targets):
        ce_loss = nn.CrossEntropyLoss(reduction='none')(inputs, targets)
        pt = torch.exp(-ce_loss)
        focal_loss = self.alpha * (1 - pt) ** self.gamma * ce_loss

        if self.reduction == 'mean':
            return focal_loss.mean()
        elif self.reduction == 'sum':
            return focal_loss.sum()
        return focal_loss

# Custom Trainer
class CustomTrainer(Trainer):
    def __init__(self, loss_fn, tokenizer, *args, **kwargs):
        super().__init__(*args, **kwargs)
        self.loss_fn = loss_fn
        self.tokenizer = tokenizer

    def compute_loss(self, model, inputs, return_outputs=False, **kwargs):
        labels = inputs.pop("labels")
        outputs = model(**inputs)
        logits = outputs.logits
        loss = self.loss_fn(logits, labels)
        return (loss, outputs) if return_outputs else loss


# Hàm encode
def encode_data(df, tokenizer):
    encodings = tokenizer(list(df['text']), truncation=True, padding=True, max_length=512, return_tensors="pt")
    return encodings

# Chuẩn bị dataset
def prepare_dataset(encodings, labels):
    return Dataset.from_dict({
        "input_ids": encodings["input_ids"],
        "attention_mask": encodings["attention_mask"],
        "labels": labels.tolist()
    })

# Đọc dữ liệu
train_df = pd.read_csv("train.csv")
val_df = pd.read_csv("val.csv")

# Cấu hình
BATCH_SIZE = 32
EPOCHS = 10

teacher_models = {
    "PhoBERT": {"model": "vinai/phobert-base", "gradient_accumulation_steps": 1},
    #"mBART": {"model": "facebook/mbart-large-cc25", "gradient_accumulation_steps": 2},
}

loss_methods = {
    "PhoBERT": FocalLoss(),
    #"mBART": FocalLoss(),
}

# Tạo thư mục models nếu chưa có
os.makedirs("./models", exist_ok=True)

completed_models = [m for m in os.listdir("./models/") if os.path.isdir(f"./models/{m}")]

torch.backends.cudnn.benchmark = True

# Huấn luyện
for model_name, config in teacher_models.items():
    if model_name in completed_models:
        print(f"\n{model_name} đã được fine-tune trước đó, bỏ qua...")
        continue

    print(f"\nFine-tuning {model_name}...")

    # Load tokenizer
    try:
        tokenizer = AutoTokenizer.from_pretrained(config["model"], trust_remote_code="qwen" in config["model"].lower())

        if tokenizer.pad_token is None:
            tokenizer.add_special_tokens({'pad_token': '[PAD]'})

        # Nếu là mBART, thêm src_lang
        if "mbart" in config["model"].lower():
            tokenizer.src_lang = "vi_VN"

    except Exception as e:
        print(f"Lỗi tải tokenizer cho {model_name}: {e}")
        continue


    # Encode dữ liệu
    train_encodings = encode_data(train_df, tokenizer)
    val_encodings = encode_data(val_df, tokenizer)

    # Chuẩn bị dataset
    train_dataset = prepare_dataset(train_encodings, train_df["label"])
    val_dataset = prepare_dataset(val_encodings, val_df["label"])

    # Load model
    try:
        model = AutoModelForSequenceClassification.from_pretrained(config["model"], num_labels=10)
        model.resize_token_embeddings(len(tokenizer))
    except Exception as e:
        print(f"Lỗi tải model cho {model_name}: {e}")
        continue

    training_args = TrainingArguments(
        output_dir=f"./results/{model_name}",
        evaluation_strategy="epoch",
        save_strategy="epoch",
        per_device_train_batch_size=BATCH_SIZE,
        per_device_eval_batch_size=BATCH_SIZE,
        num_train_epochs=EPOCHS,
        gradient_accumulation_steps=config["gradient_accumulation_steps"],
        fp16=True,
        optim="adamw_torch",
        logging_dir=f"./logs/{model_name}",
        logging_steps=100,
        save_total_limit=2,
        load_best_model_at_end=False,
        report_to="none",
    )

    trainer = CustomTrainer(
      model=model,
      args=training_args,
      train_dataset=train_dataset,
      eval_dataset=val_dataset,
      tokenizer=tokenizer,
      loss_fn=loss_methods[model_name],
      compute_metrics=compute_metrics
  )

    try:
        start_train_time = time.time()
        trainer.train()
        end_train_time = time.time()
        train_times[model_name] = end_train_time - start_train_time
    except torch.cuda.OutOfMemoryError:
        print(f"GPU bị tràn bộ nhớ khi training {model_name}")
        torch.cuda.empty_cache()
        train_times[model_name] = -1

    # Lưu model
    model.save_pretrained(f"./models/{model_name}")
    tokenizer.save_pretrained(f"./models/{model_name}")
    print(f"\nModel {model_name} fine-tuned and saved!")

    del model
    del trainer
    torch.cuda.empty_cache()
```

```python
import torch
import pandas as pd
import os
import numpy as np
from transformers import AutoModelForSequenceClassification, AutoTokenizer
from scipy.spatial.distance import mahalanobis

# Load mô hình tốt nhất
model_path = "./best_model"
tokenizer = AutoTokenizer.from_pretrained(model_path)
model = AutoModelForSequenceClassification.from_pretrained(model_path)
model.eval()

# Load test dataset
test_df = pd.read_csv("test.csv")

def encode_data(df, tokenizer):
    return tokenizer(list(df['text']), truncation=True, padding=True, max_length=512, return_tensors="pt")

# **1Xác định trung tâm dữ liệu train (để tính Mahalanobis)**
train_df = pd.read_csv("train.csv")
train_encodings = encode_data(train_df, tokenizer)

with torch.no_grad():
    train_outputs = model(**train_encodings)
    train_logits = train_outputs.logits.numpy()

# Tính trung bình & ma trận hiệp phương sai
train_mean = np.mean(train_logits, axis=0)
train_cov = np.cov(train_logits.T)
train_inv_cov = np.linalg.inv(train_cov)  # Ma trận nghịch đảo

# **2Xử lý dữ liệu test**
test_encodings = encode_data(test_df, tokenizer)

with torch.no_grad():
    test_outputs = model(**test_encodings)
    test_logits = test_outputs.logits.numpy()

# **3Kiểm tra OOD bằng Score Threshold + Mahalanobis Distance**
THRESHOLD = 0.8
MAHALANOBIS_THRESHOLD = 5.0

ood_results = []
for i, logit in enumerate(test_logits):
    prob = torch.softmax(torch.tensor(logit), dim=-1)
    max_score = torch.max(prob).item()
    distance = mahalanobis(logit, train_mean, train_inv_cov)

    if max_score < THRESHOLD or distance > MAHALANOBIS_THRESHOLD:
        ood_results.append({"text": test_df["text"][i], "OOD_Reason": "Low Score" if max_score < THRESHOLD else "High Mahalanobis Distance"})

# **Lưu các mẫu OOD**
if ood_results:
    ood_df = pd.DataFrame(ood_results)
    ood_df.to_csv("ood_samples.csv", index=False)
    print(f"\nFound {len(ood_results)} OOD samples! Saved to ood_samples.csv.")
else:
    print("\nNo OOD samples detected.")
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
from transformers import AutoModel, AutoConfig, AutoTokenizer
from torch.utils.data import TensorDataset, DataLoader
import pandas as pd
```

```python
class SmartStudentPhoBERT(nn.Module):
    def __init__(self, pretrained_model="vinai/phobert-base", num_labels=10, num_student_layers=4):
        super(SmartStudentPhoBERT, self).__init__()
        config = AutoConfig.from_pretrained(pretrained_model, output_hidden_states=True)
        self.config = config
        full_model = AutoModel.from_pretrained(pretrained_model, config=config)
        full_model.encoder.layer = nn.ModuleList(full_model.encoder.layer[:num_student_layers])
        self.backbone = full_model
        hidden_size = config.hidden_size

        self.attn_pool = nn.Sequential(nn.Linear(hidden_size, 1), nn.Softmax(dim=1))
        self.gate = nn.Sequential(nn.Linear(hidden_size * 2, hidden_size), nn.Sigmoid())
        self.projection = nn.Sequential(nn.Linear(hidden_size, 512), nn.ReLU(), nn.Dropout(0.1))
        self.classifier = nn.Linear(512, num_labels)

    def forward(self, input_ids, attention_mask, output_attentions=False, output_hidden_states=False):
        outputs = self.backbone(input_ids=input_ids, attention_mask=attention_mask,
                                output_attentions=output_attentions, output_hidden_states=output_hidden_states)
        hidden_states = outputs.last_hidden_state
        cls_token = hidden_states[:, 0, :]
        attn_scores = self.attn_pool(hidden_states)
        pooled_token = torch.sum(attn_scores * hidden_states, dim=1)
        fusion = torch.cat([cls_token, pooled_token], dim=1)
        gate = self.gate(fusion)
        fused = gate * cls_token + (1 - gate) * pooled_token
        proj = self.projection(fused)
        logits = self.classifier(proj)
        return {"logits": logits, "cls_rep": fused,
                "attentions": outputs.attentions if output_attentions else None,
                "hidden_states": outputs.hidden_states if output_hidden_states else None}
```

```python
#Cài đặt các Loss Function

import torch
import torch.nn as nn
import torch.nn.functional as F

# General Distillation Loss
class GeneralDistillationLoss(nn.Module):
    def __init__(self, alpha=0.7, T=2.0, loss_type='kl'):
        super().__init__()
        self.alpha = alpha
        self.T = T
        self.ce = nn.CrossEntropyLoss()
        self.loss_type = loss_type.lower()

    def forward(self, student_logits, teacher_logits, labels):
        soft_student = F.log_softmax(student_logits / self.T, dim=-1)
        soft_teacher = F.softmax(teacher_logits / self.T, dim=-1)

        if self.loss_type == 'kl':
            soft_loss = F.kl_div(soft_student, soft_teacher, reduction="batchmean") * self.T**2
        elif self.loss_type == 'mse':
            soft_loss = F.mse_loss(student_logits, teacher_logits)
        elif self.loss_type == 'cosine':
            soft_loss = 1 - F.cosine_similarity(student_logits, teacher_logits, dim=-1).mean()
        elif self.loss_type == 'jsd':
            m = 0.5 * (soft_teacher + torch.exp(soft_student))
            soft_loss = 0.5 * (F.kl_div(soft_student, m, reduction='batchmean') +
                               F.kl_div(torch.log(m), soft_teacher, reduction='batchmean'))
        else:
            raise ValueError("Unsupported loss type")

        hard_loss = self.ce(student_logits, labels)
        return self.alpha * soft_loss + (1 - self.alpha) * hard_loss
```

```python
class GeneralDistillationLoss(nn.Module):
    def __init__(self, alpha=0.7, T=2.0, loss_type='kl'):
        super().__init__()
        self.alpha = alpha
        self.T = T
        self.ce = nn.CrossEntropyLoss()
        self.loss_type = loss_type.lower()

    def forward(self, student_logits, teacher_logits, labels):
        soft_student = F.log_softmax(student_logits / self.T, dim=-1)
        soft_teacher = F.softmax(teacher_logits / self.T, dim=-1)
        if self.loss_type == 'kl':
            soft_loss = F.kl_div(soft_student, soft_teacher, reduction="batchmean") * self.T**2
        elif self.loss_type == 'mse':
            soft_loss = F.mse_loss(student_logits, teacher_logits)
        elif self.loss_type == 'cosine':
            soft_loss = 1 - F.cosine_similarity(student_logits, teacher_logits, dim=-1).mean()
        elif self.loss_type == 'jsd':
            m = 0.5 * (soft_teacher + torch.exp(soft_student))
            soft_loss = 0.5 * (F.kl_div(soft_student, m, reduction='batchmean') +
                               F.kl_div(torch.log(m), soft_teacher, reduction='batchmean'))
        else:
            raise ValueError("Unsupported loss type")
        hard_loss = self.ce(student_logits, labels)
        return self.alpha * soft_loss + (1 - self.alpha) * hard_loss
```

```python
class SupConLoss(nn.Module):
    def __init__(self, temperature=0.07):
        super(SupConLoss, self).__init__()
        self.temperature = temperature

    def forward(self, features, labels):
        device = features.device
        labels = labels.contiguous().view(-1, 1)
        mask = torch.eq(labels, labels.T).float().to(device)
        contrast_feature = features
        anchor_dot_contrast = torch.div(torch.matmul(contrast_feature, contrast_feature.T), self.temperature)
        logits_max, _ = torch.max(anchor_dot_contrast, dim=1, keepdim=True)
        logits = anchor_dot_contrast - logits_max.detach()
        exp_logits = torch.exp(logits) * (1 - torch.eye(labels.shape[0]).to(device))
        log_prob = logits - torch.log(exp_logits.sum(1, keepdim=True) + 1e-12)
        mean_log_prob_pos = (mask * log_prob).sum(1) / mask.sum(1)
        return -mean_log_prob_pos.mean()

class CenterLoss(nn.Module):
    def __init__(self, num_classes, feat_dim):
        super(CenterLoss, self).__init__()
        self.centers = nn.Parameter(torch.randn(num_classes, feat_dim))

    def forward(self, features, labels):
        device = self.centers.device
        features = features.to(device)
        labels = labels.to(device)
        centers_batch = self.centers.index_select(0, labels)
        return F.mse_loss(features, centers_batch)

def triplet_loss(embeddings, labels, margin=1.0):
    pairwise_dist = torch.cdist(embeddings, embeddings, p=2)
    labels = labels.unsqueeze(1)
    mask_anchor_positive = (labels == labels.T).float()
    mask_anchor_negative = (labels != labels.T).float()
    anchor_positive_dist = pairwise_dist * mask_anchor_positive
    hardest_positive_dist, _ = anchor_positive_dist.max(dim=1)
    anchor_negative_dist = pairwise_dist + 1e5 * mask_anchor_positive
    hardest_negative_dist, _ = anchor_negative_dist.min(dim=1)
    return F.relu(hardest_positive_dist - hardest_negative_dist + margin).mean()
```

```python
def train_one_epoch(model, dataloader, optimizer, device, teacher=None,
                    loss_type="jsd", metric_loss_type=None, alpha=0.7, T=2.0,
                    supcon_loss=None, center_loss=None):

    model.train()
    total_loss, total_correct, total_samples = 0.0, 0, 0
    distill_loss_fn = GeneralDistillationLoss(alpha=alpha, T=T, loss_type=loss_type)

    for batch in dataloader:
        input_ids, attention_mask, labels = [t.to(device) for t in batch]

        with torch.no_grad():
            teacher_logits = teacher(input_ids=input_ids, attention_mask=attention_mask).logits if teacher else None

        student_output = model(input_ids=input_ids, attention_mask=attention_mask)
        logits = student_output["logits"]
        embedding = student_output.get("cls_rep", None)

        if teacher_logits is not None:
            loss = distill_loss_fn(logits, teacher_logits, labels)
        else:
            loss = nn.CrossEntropyLoss()(logits, labels)

        if embedding is not None:
            embedding = F.normalize(embedding, dim=-1)
            if metric_loss_type == "supcon":
                loss += supcon_loss(embedding, labels)
            elif metric_loss_type == "center":
                loss += center_loss(embedding, labels)
            elif metric_loss_type == "triplet":
                loss += triplet_loss(embedding, labels)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        total_loss += loss.item()
        preds = torch.argmax(logits, dim=-1)
        total_correct += (preds == labels).sum().item()
        total_samples += labels.size(0)

    return total_loss / len(dataloader), total_correct / total_samples
```

```python
import pandas as pd
import torch
from torch.utils.data import TensorDataset, DataLoader
from transformers import AutoTokenizer

# Đọc dữ liệu đã có
train_df = pd.read_csv("train.csv")

# Dùng đúng tokenizer của PhoBERT
tokenizer = AutoTokenizer.from_pretrained("vinai/phobert-base")

def encode_data(df, tokenizer, max_length=256):
    encodings = tokenizer(
        list(df['text']),
        truncation=True,
        padding=True,
        max_length=max_length,
        return_tensors="pt"
    )
    return encodings

# Encode
train_encodings = encode_data(train_df, tokenizer)
train_labels = torch.tensor(train_df["label"].values)

# Tạo TensorDataset và DataLoader
train_dataset = TensorDataset(
    train_encodings["input_ids"],
    train_encodings["attention_mask"],
    train_labels
)

BATCH_SIZE = 16
train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True)
```

```python
#Hàm tính Agreement Score & Cosine Token Importance
from sklearn.metrics import accuracy_score
import torch.nn.functional as F
import numpy as np

def agreement_score(logits1, logits2):
    pred1 = torch.argmax(logits1, dim=-1)
    pred2 = torch.argmax(logits2, dim=-1)
    return accuracy_score(pred1.cpu(), pred2.cpu())

def cosine_token_importance(logits1, logits2):
    return F.cosine_similarity(logits1, logits2, dim=-1).mean().item()

```

```python
#LIME: Tính unmatched samples giữa student và teacher
from lime.lime_text import LimeTextExplainer
import numpy as np

explainer = LimeTextExplainer(class_names=[str(i) for i in range(10)])
threshold_tau = 0.1  # Ngưỡng LIME diff để đánh dấu unmatched

def lime_predict(model, tokenizer, texts, device):
    inputs = tokenizer(texts, return_tensors='pt', padding=True, truncation=True, max_length=512)
    inputs = {k: v.to(device) for k, v in inputs.items()}

    # Xóa token_type_ids nếu model không support
    if 'token_type_ids' in inputs:
        del inputs['token_type_ids']

    with torch.no_grad():
        logits = model(**inputs)["logits"] if isinstance(model, SmartStudentPhoBERT) else model(**inputs).logits
        probs = torch.softmax(logits, dim=-1).cpu().numpy()
    return probs
```

```python
#Hàm phân tích LIME và thu thập unmatched samples
def lime_analysis(texts, labels, teacher_model, student_model, tokenizer, device):
    unmatched_summary = []
    for i, text in enumerate(texts):
        label = labels[i]
        teacher_pred = np.argmax(lime_predict(teacher_model, tokenizer, [text], device))
        student_pred = np.argmax(lime_predict(student_model, tokenizer, [text], device))

        exp_teacher = explainer.explain_instance(text, lambda x: lime_predict(teacher_model, tokenizer, x, device), num_features=5)
        exp_student = explainer.explain_instance(text, lambda x: lime_predict(student_model, tokenizer, x, device), num_features=5)

        words_teacher = dict(exp_teacher.as_list())
        words_student = dict(exp_student.as_list())

        common_words = set(words_teacher.keys()) & set(words_student.keys())
        if not common_words:
            avg_diff = 1.0
        else:
            avg_diff = np.mean([abs(words_teacher[w] - words_student[w]) for w in common_words])

        if avg_diff > threshold_tau or teacher_pred != student_pred:
            diff_words = sorted(set(words_teacher.keys()).symmetric_difference(set(words_student.keys())))
            unmatched_summary.append({
                "text": text,
                "label": label,
                "student_pred": student_pred,
                "teacher_pred": teacher_pred,
                "avg_lime_diff": avg_diff,
                "top_mismatched_words": ", ".join(diff_words)
            })

    return unmatched_summary
```

```python
 #Hàm thống kê unmatched theo class (heatmap data)
from collections import Counter

def count_unmatched_by_class(unmatched_summary):
    class_counts = Counter()
    for sample in unmatched_summary:
        class_counts[sample["label"]] += 1
    return class_counts
```

```python
#SHAP: Hàm tính Explainability Scores
import shap

def compute_shap_scores(model, tokenizer, texts, device, background_size=50):
    # Đảm bảo texts là List[str]
    texts = list(map(str, texts))

    background_texts = texts[:background_size]
    tokenizer_kwargs = {"max_length": 512, "padding": True, "truncation": True, "return_tensors": "pt"}

    def predict(texts):
        texts = list(map(str, texts))
        inputs = tokenizer(texts, **tokenizer_kwargs)

        if 'token_type_ids' in inputs:
            del inputs['token_type_ids']

        inputs = {k: v.to(device) for k, v in inputs.items()}
        with torch.no_grad():
            outputs = model(**inputs)
            logits = outputs["logits"] if isinstance(outputs, dict) else outputs.logits
        return torch.softmax(logits, dim=-1).cpu().numpy()

    explainer = shap.Explainer(predict, masker=shap.maskers.Text(tokenizer))
    shap_values = explainer(texts)
    sample_scores = [np.mean(np.abs(v.values)) for v in shap_values]
    return sample_scores
```

```python
#Tính tương quan Explainability và Performance
import pandas as pd

def summarize_explainability_results(loss_type, shap_scores, unmatched_summary, acc, f1):
    avg_shap = np.mean(shap_scores)
    std_shap = np.std(shap_scores)
    unmatched_count = len(unmatched_summary)

    return {
        "Loss": loss_type,
        "Avg_SHAP": avg_shap,
        "Std_SHAP": std_shap,
        "Unmatched": unmatched_count,
        "Accuracy": acc,
        "F1": f1
    }
```

```python
#Update 1
def compute_top_diff_samples(texts, labels, unmatched_indices, teacher_imps, student_imps, tokenized_texts,
                             teacher_model, student_model, tokenizer, device, top_k=10):
    import numpy as np
    import pandas as pd

    rows = []
    for idx in unmatched_indices:
        t_imp = np.array(teacher_imps[idx])
        s_imp = np.array(student_imps[idx])
        tokens = tokenized_texts[idx]

        diff = np.abs(t_imp - s_imp)
        avg_diff = diff.mean()

        # top n token gây khác biệt
        top_token_ids = np.argsort(diff)[-5:]
        top_tokens = [tokens[i] for i in top_token_ids]

        input_enc = tokenizer(texts[idx], return_tensors="pt", truncation=True, padding=True, max_length=512).to(device)

        with torch.no_grad():
            t_pred = torch.argmax(teacher_model(**input_enc).logits, dim=1).item()
            s_pred = torch.argmax(student_model(**input_enc).logits, dim=1).item()

        rows.append({
            "Text": texts[idx],
            "True Label": labels[idx],
            "Teacher Pred": t_pred,
            "Student Pred": s_pred,
            "Avg_LIME_Diff": round(avg_diff, 4),
            "Top_Mismatched_Tokens": ", ".join(top_tokens)
        })

    df = pd.DataFrame(rows)
    df = df.sort_values("Avg_LIME_Diff", ascending=False).head(top_k)
    return df
```

```python
#Update 2
def evaluate_explainability_pipeline(loss_type, student_model_path, teacher_model, tokenizer, test_texts, test_labels, acc, f1, device,
                                     run_lime=True, run_shap=True):
    student_model = load_student_model(student_model_path, device)

    unmatched, lime_class_counts, shap_scores = [], {}, []
    top_diff_df = None

    if run_lime:
        print(f"  ➤ Running LIME analysis for {loss_type}...")
        unmatched, teacher_importances, student_importances, tokenized_texts = lime_analysis(
            test_texts, test_labels, teacher_model, student_model, tokenizer, device
        )
        lime_class_counts = count_unmatched_by_class(unmatched)

        top_diff_df = compute_top_diff_samples(
            test_texts, test_labels, unmatched,
            teacher_importances, student_importances, tokenized_texts,
            teacher_model, student_model, tokenizer, device
        )

    if run_shap:
        print(f"  ➤ Running SHAP analysis for {loss_type}...")
        shap_scores = compute_shap_scores(student_model, tokenizer, test_texts, device)

    print(f"  ➤ Summarizing explainability results for {loss_type}...")
    summary = summarize_explainability_results(loss_type, shap_scores, unmatched, acc, f1)

    return summary, unmatched, lime_class_counts, shap_scores, top_diff_df
```

```python
#Feature Alignment Loss Function
import torch.nn.functional as F

def feature_alignment_loss(student_feat, teacher_feat, loss_type='mse'):
    if loss_type == 'mse':
        return F.mse_loss(student_feat, teacher_feat)
    elif loss_type == 'cosine':
        return 1 - F.cosine_similarity(student_feat, teacher_feat, dim=-1).mean()
    else:
        raise ValueError("Unsupported alignment loss type: choose 'mse' or 'cosine'")
```

```python
#Fine-tune Epochs với Feature Alignment
def fine_tune_with_alignment(student_model, teacher_model, dataloader, device,
                             align_loss_type='mse', align_weight=1.0, epochs=3, lr=2e-5):

    optimizer = torch.optim.AdamW(student_model.parameters(), lr=lr)
    student_model.train()
    teacher_model.eval()

    for epoch in range(epochs):
        total_loss = 0.0
        total_samples = 0

        for batch in dataloader:
            input_ids, attention_mask, labels = [b.to(device) for b in batch]

            with torch.no_grad():
                teacher_outputs = teacher_model(
                    input_ids=input_ids,
                    attention_mask=attention_mask,
                    output_hidden_states=True
                )
                teacher_embed = teacher_outputs.hidden_states[-1][:, 0, :]  # CLS token

            student_outputs = student_model(
                input_ids=input_ids,
                attention_mask=attention_mask,
                output_attentions=False,
                output_hidden_states=False
            )

            if isinstance(student_outputs, dict):
                student_embed = student_outputs.get("cls_rep", None)
            else:
                raise ValueError("Student model must return 'cls_rep' for alignment.")

            loss = feature_alignment_loss(student_embed, teacher_embed, loss_type=align_loss_type)
            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

            total_loss += loss.item()
            total_samples += 1

        avg_loss = total_loss / total_samples
        print(f"[Align-Epoch {epoch+1}/{epochs}] Feature Alignment Loss: {avg_loss:.4f}")
```

```python
from sklearn.metrics import accuracy_score, f1_score, precision_score, recall_score, confusion_matrix
import torch.nn.functional as F
import numpy as np
import time

def evaluate_student(model, dataloader, device):
    model.eval()
    y_true, y_pred = [], []
    inference_times = []

    with torch.no_grad():
        for batch in dataloader:
            input_ids, attention_mask, labels = [b.to(device) for b in batch]

            start = time.time()
            outputs = model(input_ids=input_ids, attention_mask=attention_mask)
            end = time.time()

            logits = outputs.logits
            preds = torch.argmax(logits, dim=-1)

            y_true.extend(labels.cpu().numpy())
            y_pred.extend(preds.cpu().numpy())
            inference_times.append(end - start)

    return {
        "accuracy": accuracy_score(y_true, y_pred),
        "f1": f1_score(y_true, y_pred, average='weighted'),
        "precision": precision_score(y_true, y_pred, average='weighted'),
        "recall": recall_score(y_true, y_pred, average='weighted'),
        "inference_time": np.mean(inference_times),
        "y_true": y_true,
        "y_pred": y_pred
    }
```

```python
#So sánh alignment với teacher (agreement, cosine)
def agreement_score(student_logits, teacher_logits):
    pred_s = torch.argmax(student_logits, dim=-1)
    pred_t = torch.argmax(teacher_logits, dim=-1)
    return (pred_s == pred_t).float().mean().item()

def cosine_similarity_logits(student_logits, teacher_logits):
    return F.cosine_similarity(student_logits, teacher_logits, dim=-1).mean().item()
```

```python
# LIME: Local Explainability & unmatched samples
from lime.lime_text import LimeTextExplainer
import numpy as np

explainer = LimeTextExplainer(class_names=[str(i) for i in range(10)])
tau = 0.1

def lime_predict(model, tokenizer, texts, device):
    inputs = tokenizer(texts, return_tensors='pt', padding=True, truncation=True, max_length=512)
    inputs = {k: v.to(device) for k, v in inputs.items()}
    with torch.no_grad():
        logits = model(**inputs).logits
        probs = torch.softmax(logits, dim=-1).cpu().numpy()
    return probs

def lime_analysis(texts, labels, teacher_model, student_model, tokenizer, device):
    unmatched_summary = []
    for i, text in enumerate(texts):
        label = labels[i]
        teacher_pred = np.argmax(lime_predict(teacher_model, tokenizer, [text], device))
        student_pred = np.argmax(lime_predict(student_model, tokenizer, [text], device))

        exp_teacher = explainer.explain_instance(text, lambda x: lime_predict(teacher_model, tokenizer, x, device), num_features=5)
        exp_student = explainer.explain_instance(text, lambda x: lime_predict(student_model, tokenizer, x, device), num_features=5)

        words_teacher = dict(exp_teacher.as_list())
        words_student = dict(exp_student.as_list())

        common_words = set(words_teacher.keys()) & set(words_student.keys())
        avg_diff = np.mean([abs(words_teacher[w] - words_student[w]) for w in common_words]) if common_words else 1.0

        if avg_diff > tau or teacher_pred != student_pred:
            diff_words = sorted(set(words_teacher.keys()).symmetric_difference(set(words_student.keys())))
            unmatched_summary.append({
                "text": text,
                "label": label,
                "student_pred": student_pred,
                "teacher_pred": teacher_pred,
                "avg_lime_diff": avg_diff,
                "top_mismatched_words": ", ".join(diff_words)
            })
    return unmatched_summary

# Gọi hàm
unmatched_summary = lime_analysis(test_texts, test_labels, teacher_model, aligned_model, tokenizer, device)
```

```python
#Heatmap số unmatched theo class
from collections import Counter
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

lime_class_counts = Counter()
for s in unmatched_summary:
    lime_class_counts[s["label"]] += 1

heatmap_df = pd.DataFrame.from_dict(lime_class_counts, orient="index", columns=["Unmatched"]).sort_index()

plt.figure(figsize=(6, 5))
sns.heatmap(heatmap_df.T, annot=True, fmt="d", cmap="YlGnBu")
plt.title("Unmatched Samples per Class (LIME) - After Alignment")
plt.xlabel("Class Label")
plt.tight_layout()
plt.show()
```

```python
#SHAP: Global Explainability Score
import shap

def compute_shap_scores(model, tokenizer, texts, device, background_size=50):
    background_texts = texts[:background_size]
    tokenizer_kwargs = {"max_length": 512, "padding": True, "truncation": True, "return_tensors": "pt"}

    def predict(texts):
        inputs = tokenizer(texts, **tokenizer_kwargs).to(device)
        with torch.no_grad():
            logits = model(**inputs).logits
        return torch.softmax(logits, dim=-1).cpu().numpy()

    explainer = shap.Explainer(predict, masker=shap.maskers.Text(tokenizer))
    shap_values = explainer(texts)
    sample_scores = [np.mean(np.abs(val.values)) for val in shap_values]
    return sample_scores

shap_scores = compute_shap_scores(aligned_model, tokenizer, test_texts, device)
```
